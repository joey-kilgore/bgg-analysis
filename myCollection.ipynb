```python
me_and_mickey = []
for p in plays:
    if len(p.players) > 2 and "Joey" in p.players and "Mickey" in p.players:
        me_and_mickey.append(p)

num_mickey_win = 0
num_joey_win = 0
num_total_wins = 0
skipped_games = ["Dominion (Second Edition) Big Box","Dominion"]
# skipped_games = []
for p in me_and_mickey:
    if p.game in skipped_games:
        continue

    if "Mickey" in p.winningPlayers:
        num_mickey_win += 1
    if "Joey" in p.winningPlayers:
        num_joey_win += 1
    if p.winningPlayers == []:
        print(f"{p.game} NO WINNER")
    elif "Joey" in p.winningPlayers and "Mickey" in p.winningPlayers:
        print(f"{p.game} BOTH WON")
        num_joey_win -= 1
        num_mickey_win -= 1
    else:
        num_total_wins += 1

print("JOEY AND MICKEY AND OTHERS")
print(f"TOTAL WINS = {num_total_wins}")
print(f"MICKEY WINS = {num_mickey_win}")
print(f"JOEY WINS : {num_joey_win}")
print(f"MICKEY WIN PERCENTAGE : {(num_mickey_win/(num_total_wins))*100:0.2f}")
print(f"JOEY WIN PERCENTAGE : {(num_joey_win/(num_total_wins))*100:0.2f}")
```
